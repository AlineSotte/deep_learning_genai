### 1. **Qual é a principal função de uma camada convolucional em uma CNN?**

- A) Normalizar os dados
- B) Detectar padrões e características
- C) Reduzir a dimensionalidade dos dados
- D) Aumentar a resolução da imagem


Sua resposta: B


### 2. **O que é uma operação de pooling em uma CNN?**

- A) Uma técnica para aumentar a complexidade do modelo
- B) Uma função de ativação
- C) Um método para reduzir a dimensionalidade dos mapas de características
- D) Um algoritmo de otimização


Sua resposta: C


### 3. **Qual é a função da camada de ativação ReLU em uma CNN?**

- A) Introduzir não-linearidade no modelo
- B) Reduzir o overfitting
- C) Normalizar os dados de entrada
- D) Aumentar a precisão do modelo


Sua resposta: A


### 4. **O que é o "stride" em uma operação convolucional?**

- A) O tamanho do kernel de convolução
- B) O número de camadas convolucionais
- C) O passo de movimentação do kernel sobre os dados de entrada
- D) O número de épocas de treinamento


Sua resposta: C


### 5. **Qual é o propósito da técnica de "padding" em uma CNN?**

- A) Aumentar a complexidade do modelo
- B) Melhorar a eficiência computacional
- C) Reduzir o número de parâmetros
- D) Preservar as dimensões espaciais dos dados de entrada


Sua resposta: D


### 6. **Qual das seguintes opções é uma técnica para prevenir overfitting em CNNs?**

- A) Aumentar o número de camadas
- B) Usar a técnica de dropout
- C) Reduzir o tamanho do conjunto de dados
- D) Aumentar a taxa de aprendizado


Sua resposta: B


### 7. **Qual é a principal vantagem de usar modelos pré-treinados em CNNs?**

- A) Reduzir o tempo de treinamento
- B) Aumentar o tamanho do modelo
- C) Melhorar a precisão em novas tarefas
- D) Reduzir o uso de memória


Sua resposta: A


### 8. **O que é uma camada totalmente conectada (fully connected) em uma CNN?**

- A) Uma camada que conecta cada neurônio de uma camada à próxima camada
- B) Uma camada que realiza operações de pooling
- C) Uma camada que normaliza os dados
- D) Uma camada que aplica funções de ativação

- C) Adicionar regularização ao modelo
- D) Estabilizar o treinamento e mitigar gradientes explosivos ou desaparecidos

### 10. **Qual técnica de preprocessamento permite que as CNNs processem imagens de tamanhos variados?**
