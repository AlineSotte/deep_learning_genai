### 1. **O que é transferência de aprendizado em CNNs?**

- A) Treinar uma rede neural do zero
- B) Usar um modelo pré-treinado em uma nova tarefa
- C) Aumentar o número de camadas em uma rede
- D) Reduzir o tamanho do conjunto de dados


Sua resposta: B


### 2. **Qual é a principal vantagem de usar transferência de aprendizado em CNNs?**

- A) Reduzir o tempo de treinamento
- B) Aumentar a complexidade do modelo
- C) Necessitar de mais dados de treinamento
- D) Melhorar a precisão em tarefas não relacionadas

Sua resposta: A
